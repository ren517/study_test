### 第四题

```python
def binary_search(numbers, l, r, value):
    mid = (l + r) // 2
    if l >= r:
        if l < len(numbers) and numbers[l] == value:
            return l
        return -1

    mid = (l + r) // 2
    if value == numbers[mid]:
        return mid
    elif value > numbers[mid]:
        return binary_search(numbers, mid, r, value)
    else:
        return binary_search(numbers, l, mid, value)


numbers = (-10, 1, 5, 8, 9, 13, 17, 19, 20, 24, 29, 31, 33, 45, 67, 88, 89, 90, 100)
print(binary_search(numbers, 0, len(numbers), -100))
```

### 第六题

```python
import random
from operator import itemgetter

scores = [[id, random.randrange(50, 101)] for id in range(1, 31)]
new_scores = sorted(scores, key=itemgetter(1), reverse=True)
for i in range(len(new_scores)):
    print(f"id : {new_scores[i][0]}  score : {new_scores[i][1]}, 排名：{i + 1}")
```
